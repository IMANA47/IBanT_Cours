```python
""" Exercice

Un club privé n'autorise l’entrée qu’à certaines conditions :

- La personne doit avoir au moins 18 ans.
- Si elle a entre 18 et 25 ans, elle peut entrer gratuitement
- Si elle a plus de 25 ans, elle doit être soit membre du club, soit accompagnée d’un membre.
- Toute autre situation est refusée."""


age = 0
membre = False
accompagne = True

#- La personne doit avoir au moins 18 ans.
if age >= 18:

    #- Si elle a entre 18 et 25 ans, elle peut entrer gratuitement
    if 18 <= age <= 25:
        print("Accès autorisé : vous pouvez entrer gratuitement.")
    
    # - Si elle a plus de 25 ans, elle doit être soit membre du club, soit accompagnée d’un membre.
    elif age > 25:
        if membre or accompagne:
            print("Accès autorisé : vous êtes membre du club ou accompagné d'un membre.")
        else:
            print("Accès refusé : vous devez être membre du club ou accompagné d'un membre.")

# - Toute autre situation est refusée.
else:
    print("Accès refusé : vous devez avoir au moins 18 ans.")
```
