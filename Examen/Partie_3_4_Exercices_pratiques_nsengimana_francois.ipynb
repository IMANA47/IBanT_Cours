Exercice 2:
Creer une fonction : calcul_moyenne(notes) qui recoit une liste de notes et retourne:
- La moyenne;
- La meilleure note;
- La plus faible note

```python
notes = [12, 15, 9, 18, 14]
def calcul_moyenne(notes):
    moyenne = sum(notes) / len(notes)
    meilleure_note = max(notes)
    plus_faible_note = min(notes)
    print(f"La moyenne : {moyenne}")
    print(f"La meilleure note est: {meilleure_note}")
    print(f"La plus faible note est: {plus_faible_note}")

print(calcul_moyenne(notes))
```

Partie 4 : Mini-Projet 
Gestion d'une bibliotheque : Creer un programme oriente objet permettant de gerer une petit bibliotheque.

```python
"""
Fonctionnalités demandees:
A- Creer une classe Livre avec les attributs:
 Titre, 
 Auteur, 
 Année,
 Disponible.
"""
class Livre:
    def __init__(self, titre, auteur, annee, disponible=True):
        self.titre = titre
        self.auteur = auteur
        self.annee = annee
        self.disponible = disponible

"""
Créez une classe Bibliotheque qui contient une liste de livres et les méthodes suivantes :
- ajouter,
- afficher ,
- rechercher, 
- emprunter,
- retourner, 
- quitter
"""
class Bibliotheque:
    def __init__(self):
        self.livres = []

    def ajouter_livre(self, livre):
        self.livres.append(livre)
        print(f"Le livre '{livre.titre}' a été ajouté à la bibliothèque.")

    def afficher_livres(self):
        if not self.livres:
            print("Aucun livre dans la bibliothèque pour le moment.")
            return

        print("Voici la liste des livres :")
        for livre in self.livres:
            statut = "Disponible" if livre.disponible else "Emprunté"
            print(f"- {livre.titre} | {livre.auteur} | {livre.annee} | {statut}")

    def rechercher(self, mot):
        resultats = []
        for livre in self.livres:
            if mot.lower() in livre.titre.lower() or mot.lower() in livre.auteur.lower():
                resultats.append(livre)

        if not resultats:
            print("Aucun livre trouvé.")
            return

        print("Résultat de la recherche de votre livre :")
        for livre in resultats:
            statut = "Disponible" if livre.disponible else "Emprunté"
            print(f"- {livre.titre} | {livre.auteur} | {livre.annee} | {statut}")

    def emprunter_livre(self, titre):
        for livre in self.livres:
            if livre.titre.lower() == titre.lower():
                if livre.disponible:
                    livre.disponible = False
                    print(f"Vous avez emprunté '{livre.titre}'.")
                else:
                    print(f"'{livre.titre}' est déjà emprunté.")
                return

        print("Livre introuvable pour emprunter.")

    def retourner_livre(self, titre):
        for livre in self.livres:
            if livre.titre.lower() == titre.lower():
                if not livre.disponible:
                    livre.disponible = True
                    print(f"Vous avez retourné '{livre.titre}'.")
                else:
                    print(f"'{livre.titre}' n'est pas emprunté.")
                return

        print("Livre introuvable pour retourner.")

    def quitter(self):
        print("Merci d'avoir utilisé la bibliothèque. Au revoir !")

    def interface_utilisateur(self):
        while True:
            print("\n1. Ajouter un livre")
            print("2. Afficher les livres")
            print("3. Rechercher un livre")
            print("4. Emprunter un livre")
            print("5. Retourner un livre")
            print("6. Quitter")

            choix_de_utilisateur = input("Choisissez une option : ")

            if choix_de_utilisateur == "1":
                titre = input("Titre : ")
                auteur = input("Auteur : ")
                annee = input("Année : ")
                livre = Livre(titre, auteur, annee)
                self.ajouter_livre(livre)

            elif choix_de_utilisateur == "2":
                self.afficher_livres()

            elif choix_de_utilisateur == "3":
                mot = input("Entrez un titre ou un auteur : ")
                self.rechercher(mot)

            elif choix_de_utilisateur == "4":
                titre = input("Titre du livre à emprunter : ")
                self.emprunter_livre(titre)

            elif choix_de_utilisateur == "5":
                titre = input("Titre du livre à retourner : ")
                self.retourner_livre(titre)

            elif choix_de_utilisateur == "6":
                self.quitter()
                break

            else:
                print("choix_de_utilisateur invalide.")


bibliotheque = Bibliotheque()
bibliotheque.interface_utilisateur()
```
